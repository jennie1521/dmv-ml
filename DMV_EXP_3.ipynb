```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("Telco-Customer-Churn.csv")

print("Dataset imported successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Display first 5 records
print("First 5 records:")
display(df.head())

# Display last 5 records
print("\nLast 5 records:")
display(df.tail())

# Dataset shape
print("\nDataset Shape:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Convert TotalCharges to numeric
# Invalid/blank values will become NaN
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Check missing values again
print("\nMissing values after converting TotalCharges:")
print(df.isnull().sum())
```

```python
# Fill missing numeric values with median
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with mode
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
# Count duplicate records
print("Number of duplicate records:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# Get categorical columns
categorical_columns = df.select_dtypes(include="object").columns

# Remove leading/trailing spaces
for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip()

# Standardize Gender
if "gender" in df.columns:
    df["gender"] = df["gender"].str.capitalize()

# Display unique values of categorical columns
for col in categorical_columns:
    print("\n", col)
    print(df[col].unique())
```

```python
# Columns containing Yes/No type values
yes_no_columns = [
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

for col in yes_no_columns:
    if col in df.columns:
        df[col] = df[col].replace({
            "yes": "Yes",
            "no": "No",
            "YES": "Yes",
            "NO": "No"
        })

print("Categorical values standardized successfully.")
```

```python
# Convert numeric columns
numeric_columns = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Convert Churn from Yes/No to 1/0
if "Churn" in df.columns:
    df["Churn"] = df["Churn"].map({
        "Yes": 1,
        "No": 0
    })

print("Data types after conversion:")
print(df.dtypes)
```

```python
# Numerical columns for outlier detection
outlier_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

for col in outlier_columns:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers = df[
            (df[col] < lower_bound) |
            (df[col] > upper_bound)
        ]
        
        print(f"{col}: {len(outliers)} outliers")
```

```python
for col in outlier_columns:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        df[col] = df[col].clip(
            lower_bound,
            upper_bound
        )

print("Outliers handled using IQR capping.")
```

```python
# Average monthly charges based on total charges and tenure
df["AvgMonthlyCharges"] = np.where(
    df["tenure"] > 0,
    df["TotalCharges"] / df["tenure"],
    df["MonthlyCharges"]
)

# Service columns
service_columns = [
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

# Count number of subscribed services
existing_services = [col for col in service_columns if col in df.columns]

df["TotalServices"] = 0

for col in existing_services:
    df["TotalServices"] += (
        df[col].astype(str).isin(["Yes", "DSL", "Fiber optic"])
    ).astype(int)

# Long-term customer feature
df["IsLongTermCustomer"] = (df["tenure"] >= 24).astype(int)

print("New features created successfully.")

display(df[[
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "AvgMonthlyCharges",
    "TotalServices",
    "IsLongTermCustomer"
]].head())
```

```python
from sklearn.preprocessing import StandardScaler

# Numerical features
scale_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "AvgMonthlyCharges",
    "TotalServices"
]

# Make sure only existing columns are selected
scale_columns = [
    col for col in scale_columns
    if col in df.columns
]

scaler = StandardScaler()

df_scaled = df.copy()

df_scaled[scale_columns] = scaler.fit_transform(
    df_scaled[scale_columns]
)

print("Numerical features standardized.")

display(df_scaled[scale_columns].head())
```

```python
from sklearn.model_selection import train_test_split

# Separate target
X = df.drop(columns=["Churn", "customerID"])

y = df["Churn"]

# Convert categorical variables into numerical variables
X = pd.get_dummies(X, drop_first=True)

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Save cleaned dataset
df.to_csv(
    "Telecom_Customer_Churn_Cleaned.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
print("File name: Telecom_Customer_Churn_Cleaned.csv")
```
