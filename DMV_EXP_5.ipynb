```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("City_day.csv")

print("Dataset imported successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nColumn Names:")
print(df.columns.tolist())

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Sort by date
df = df.sort_values("Date")

print("Relevant variables:")
print("Date, AQI, PM2.5, PM10, CO")

print(df[["Date", "AQI", "PM2.5", "PM10", "CO"]].head())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["AQI"], label="AQI")

plt.xlabel("Date")
plt.ylabel("AQI")
plt.title("Overall AQI Trend Over Time")
plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["PM2.5"], label="PM2.5")

plt.xlabel("Date")
plt.ylabel("PM2.5")
plt.title("PM2.5 Trend Over Time")
plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["PM10"], label="PM10")

plt.xlabel("Date")
plt.ylabel("PM10")
plt.title("PM10 Trend Over Time")
plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["CO"], label="CO")

plt.xlabel("Date")
plt.ylabel("CO")
plt.title("CO Trend Over Time")
plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Select a limited number of dates for better visualization
sample = df.head(30)

plt.figure(figsize=(12, 5))

plt.bar(
    sample["Date"].dt.strftime("%Y-%m-%d"),
    sample["AQI"]
)

plt.xlabel("Date")
plt.ylabel("AQI")
plt.title("AQI Comparison Across Dates")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

plt.boxplot(
    df["AQI"].dropna(),
    vert=True
)

plt.ylabel("AQI")
plt.title("Distribution of AQI Values")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["PM2.5"],
    df["AQI"],
    alpha=0.6
)

plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.title("Relationship Between AQI and PM2.5")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["PM10"],
    df["AQI"],
    alpha=0.6
)

plt.xlabel("PM10")
plt.ylabel("AQI")
plt.title("Relationship Between AQI and PM10")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["CO"],
    df["AQI"],
    alpha=0.6
)

plt.xlabel("CO")
plt.ylabel("AQI")
plt.title("Relationship Between AQI and CO")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["PM2.5"], label="PM2.5")
plt.plot(df["Date"], df["PM10"], label="PM10")
plt.plot(df["Date"], df["CO"], label="CO")

plt.xlabel("Date")
plt.ylabel("Pollutant Level")
plt.title("Pollutant Trends Over Time")

plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
